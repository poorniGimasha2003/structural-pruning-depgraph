```python
import torch 
import torch.nn as nn
```

```python
linear = nn.Linear(in_features=4 , out_features=3)
conv = nn.Conv2d(in_channels=3 , out_channels=8 , kernel_size=3)
bn = nn.BatchNorm2d(num_features=8)

print("Linear weight : " , linear.weight.shape)
print("Conv2d weight : " , conv.weight.shape)
print("BatchNorm2d weight : " , bn.weight.shape)
```

```python
conv1 = nn.Conv2d(3,8,kernel_size=3,padding=1)
bn1 = nn.BatchNorm2d(8)
conv2 = nn.Conv2d(8,16,kernel_size=3,padding=1)

print("conv1:", conv1.weight.shape, "-> 8 OUTPUT channels")
print("bn1:", bn1.weight.shape, "-> 8 channels")
print("conv2:", conv2.weight.shape, "-> 8 INPUT channels (2nd number)")
```

```python
# one fake colour image, 32×32
x = torch.randn(1,3,32,32)

# Wrong cut: conv1 now makes 7 channels, but bn1 and conv2 still expect 8
bad_conv1 = nn.Conv2d(3, 7, kernel_size=3, padding=1)

try:
    y = conv2(bn1(bad_conv1(x)))
except Exception as e:
    print("CRASH! ->", type(e).__name__)
    print(e)
    
```

```python
# one fake colour image, 32×32
x = torch.randn(1,3,32,32)

bad_conv1 = nn.Conv2d(3, 7, kernel_size=3, padding=1)
bn1 = nn.BatchNorm2d(7) #fixed
conv2 = nn.Conv2d(7,16,kernel_size=3,padding=1) #fixed
try:
    y = conv2(bn1(bad_conv1(x)))
    print("OK!...Shape of y : ",y.shape)
except Exception as e:
    print("CRASH! ->", type(e).__name__)
    print(e)
```
